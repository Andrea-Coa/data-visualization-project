# Bocetos (wireframes) de las tres vistas D3.js vinculadas

Bocetos de baja fidelidad (asistidos por IA, vía matplotlib) pensados para
comunicar estructura, encodings e interacciones antes de la Semana 7 -- no
es el diseño visual final. Refinar a mano o en una herramienta de diseño una
vez elegidos los encodings reales (escala de color, contornos del topomap).

**Nota:** el output de este notebook (`sketches/`) no se incluye en la
entrega de la Semana 6 -- se preferirán bocetos hechos a mano por cada
dueño de vista. Este notebook queda en el repo como generador de wireframes
de referencia.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle, FancyBboxPatch, Rectangle

REPO_ROOT = Path.cwd().resolve().parents[2]  # deliveries/week06/code -> raíz del repo
OUT_DIR = REPO_ROOT / "deliveries" / "week06" / "sketches"
OUT_DIR.mkdir(parents=True, exist_ok=True)

GREY = "#888888"
LIGHT = "#dddddd"
INK = "#222222"


def _panel(ax, title):
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.axis("off")
    ax.set_title(title, fontsize=13, fontweight="bold", loc="left")


def _button(ax, x, y, w, h, label, active=False):
    face = INK if active else "white"
    text_color = "white" if active else INK
    box = FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.05", linewidth=1.2,
                          edgecolor=INK, facecolor=face)
    ax.add_patch(box)
    ax.text(x + w / 2, y + h / 2, label, ha="center", va="center", fontsize=8, color=text_color)
```

## Vista 1 -- Topomap (Q1, Q2) -- dueña: Jimena

```python
fig, ax = plt.subplots(figsize=(6, 6))
_panel(ax, "View 1 -- Topomap (Q1, Q2) -- owner: Jimena")

_button(ax, 0.3, 8.8, 1.8, 0.6, "NS", active=True)
_button(ax, 2.2, 8.8, 1.8, 0.6, "SD")
_button(ax, 6.0, 8.8, 1.8, 0.6, "Mean", active=True)
_button(ax, 7.9, 8.8, 1.8, 0.6, "Variability")

scalp = Circle((5, 4.8), 3.2, fill=False, linewidth=2, edgecolor=INK)
ax.add_patch(scalp)
ax.plot([5], [8.0], marker="^", markersize=12, color=INK)  # nariz

rng = np.random.default_rng(0)
n = 61
angles = np.linspace(0, 2 * np.pi, n, endpoint=False)
radii = rng.uniform(0.4, 3.0, n)
xs = 5 + radii * np.cos(angles)
ys = 4.8 + radii * np.sin(angles) * 0.9
vals = rng.uniform(0, 1, n)
ax.scatter(xs, ys, c=vals, cmap="RdYlBu_r", s=90, edgecolors=INK, linewidths=0.5, zorder=3)

cb = Rectangle((8.6, 2.0), 0.4, 3.6, edgecolor=INK, facecolor=LIGHT)
ax.add_patch(cb)
ax.text(9.3, 3.8, "relative\ntheta\npower", fontsize=7, ha="left", va="center")
ax.text(8.6, 1.7, "low", fontsize=7, ha="left")
ax.text(8.6, 5.75, "high", fontsize=7, ha="left")

ax.text(0.3, 0.6, "Hover electrode -> tooltip (name, value, n subjects).\n"
                   "Click subject in View 3 -> highlight their per-electrode signature here.",
        fontsize=7.5, color=GREY)

fig.tight_layout()
fig.savefig(OUT_DIR / "sketch_topomap_view.png", dpi=150)
plt.show()
```

## Vista 2 -- Dinámica intra-grabación + moderadores (Q6, Q7) -- dueña: Andrea

```python
fig, ax = plt.subplots(figsize=(9, 5.5))
_panel(ax, "View 2 -- Intra-recording dynamics + moderators (Q6, Q7) -- owner: Andrea")

for i, label in enumerate(["Sex: All", "Age: All", "PSQI: All", "Order: All"]):
    _button(ax, 0.2 + i * 2.3, 8.9, 2.0, 0.55, label)

plot_box = Rectangle((0.8, 1.2), 8.4, 6.8, fill=False, edgecolor=INK, linewidth=1.2)
ax.add_patch(plot_box)
ax.text(5, 0.7, "epoch (0-75, 4s each, 5 min recording)", fontsize=8, ha="center")
ax.text(0.4, 4.6, "ROI theta\npower", fontsize=8, ha="center", rotation=90)

t = np.linspace(0.8, 9.2, 75)
vulnerable = 2 + 0.06 * (t - 0.8) + 0.15 * np.sin(t)
resistant = 4.2 + 0.02 * np.sin(t * 1.3)
ax.plot(t, vulnerable + 1.2, color="#c0392b", linewidth=1.8, label="vulnerable subject")
ax.plot(t, resistant + 1.2, color="#2c7fb8", linewidth=1.8, label="resistant subject")
ax.legend(loc="lower right", fontsize=8, frameon=False)

brush = Rectangle((5.5, 1.4), 2.0, 6.4, facecolor=INK, alpha=0.08, edgecolor=INK, linestyle="--")
ax.add_patch(brush)
ax.text(6.5, 7.9, "brush = zoom /\nepoch range", fontsize=7, ha="center", color=GREY)

ax.text(0.2, 0.15, "Line/Canvas panel; selecting a subject in View 3 highlights their trajectory here (Q4 signature check).",
        fontsize=7.5, color=GREY)

fig.tight_layout()
fig.savefig(OUT_DIR / "sketch_temporal_dynamics_view.png", dpi=150)
plt.show()
```

## Vista 3 -- Scatter de vulnerabilidad + convergencia (Q3, Q4, Q5) -- dueño: Paolo

```python
fig, ax = plt.subplots(figsize=(8.5, 7))
_panel(ax, "View 3 -- Vulnerability scatter + convergence (Q3, Q4, Q5) -- owner: Paolo")

for i, label in enumerate(["vs PVT (n=30)", "vs KSS (n=33)", "vs PANAS-PA (n=68)"]):
    active = i == 0
    _button(ax, 0.3 + i * 3.1, 8.9, 2.9, 0.55, label, active=active)

plot_box = Rectangle((0.8, 1.2), 8.0, 6.8, fill=False, edgecolor=INK, linewidth=1.2)
ax.add_patch(plot_box)
ax.axhline(4.6, color=LIGHT, linewidth=1)
ax.axvline(4.8, color=LIGHT, linewidth=1)
ax.text(4.8, 0.7, "delta theta variability (SD - NS), frontal/centro-temporal", fontsize=7.5, ha="center")
ax.text(0.35, 4.6, "delta PVT\nmedian RT", fontsize=8, ha="center", rotation=90)

rng = np.random.default_rng(1)
n = 30
xs = rng.normal(4.8, 1.6, n)
ys = rng.normal(4.6, 1.6, n)
sex = rng.choice(["M", "F"], n)
colors = ["#2c7fb8" if s == "M" else "#c0392b" for s in sex]
ax.scatter(xs, ys, c=colors, s=55, edgecolors=INK, linewidths=0.4, zorder=3)

brush = Rectangle((5.0, 5.0), 3.6, 2.9, facecolor=INK, alpha=0.06, edgecolor=INK, linestyle="--")
ax.add_patch(brush)
ax.text(6.8, 8.1, "brush = vulnerable\nsubgroup", fontsize=7, ha="center", color=GREY)

ax.scatter([], [], c="#2c7fb8", label="M")
ax.scatter([], [], c="#c0392b", label="F")
ax.legend(loc="lower right", fontsize=8, frameon=False, title="Gender")

ax.text(0.2, 0.15, "Selecting points here drives the linked brush on Views 1 and 2 (topomap + temporal panel).",
        fontsize=7.5, color=GREY)

fig.tight_layout()
fig.savefig(OUT_DIR / "sketch_vulnerability_scatter_view.png", dpi=150)
plt.show()
```
